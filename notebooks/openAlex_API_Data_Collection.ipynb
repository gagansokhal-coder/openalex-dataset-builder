```python
import pandas as pd
import requests
```

```python


dfs = []

for i in range(1, 101):   # Pages 1 to 100

    url = f"https://api.openalex.org/works?page={i}&per-page=100"

    response = requests.get(url)

    temp_df = pd.DataFrame(response.json()['results'])[
        ['id', 'title', 'display_name', 'publication_date', 'language']
    ]

    dfs.append(temp_df)

df = pd.concat(dfs, ignore_index=True)

```

```python
df.to_csv('openalex_papers.csv',index=False)
```
